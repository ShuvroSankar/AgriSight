### Temporal Alignment - Pivot FAOSTAT Years

```python
print("="*80)
print("STEP 4: TEMPORAL ALIGNMENT")
print("="*80)

# Load FAOSTAT USA data
fo_usa_df = spark.table("argisight_spark_workspace.bronze.fo_production_usa")

# Unpivot years from wide to long format
# Your columns are: Y2019, Y2020, Y2021, Y2022, Y2023
fo_usa_long_df = fo_usa_df.select(
    col("area_code"),
    col("area"),
    col("item_code"),
    col("item"),
    col("element_code"),
    col("element"),
    col("unit"),
    # Stack years into rows
    expr("stack(5, " +
         "2019, cast(Y2019 as double), " +
         "2020, cast(Y2020 as double), " +
         "2021, cast(Y2021 as double), " +
         "2022, cast(Y2022 as double), " +
         "2023, cast(Y2023 as double)" +
         ") as (year, value)")
).filter(col("value").isNotNull())

print(f"\n✓ FAOSTAT Long Format: {fo_usa_long_df.count()} records")
fo_usa_long_df.show(10)

# Filter for key metrics only: Production, Area harvested, Yield
key_elements = ["Production", "Area harvested", "Yield"]
fo_usa_key_df = fo_usa_long_df.filter(col("element").isin(key_elements))

print(f"\n✓ Key Metrics Only: {fo_usa_key_df.count()} records")

# Pivot to get Production, Area, Yield as columns
fo_usa_pivot_df = fo_usa_key_df.groupBy("year", "item_code", "item") \
    .pivot("element", key_elements) \
    .agg(first("value"))

# Rename columns for clarity
fo_usa_pivot_df = fo_usa_pivot_df \
    .withColumnRenamed("Production", "production_tonnes") \
    .withColumnRenamed("Area harvested", "area_harvested_ha") \
    .withColumnRenamed("Yield", "yield_kg_per_ha")

print("\n✓ FAOSTAT Pivoted by Year")
fo_usa_pivot_df.show(10, truncate=False)

# Save

fo_usa_pivot_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(
        "argisight_spark_workspace.bronze.fo_usa_pivot_df"
    )



print("\n✓ Saved to Delta table: argisight_spark_workspace.bronze.fo_usa_pivot_df")
print("="*80)
```

### Aggregate US Sales to Yearly

```python
from pyspark.sql.functions import *

print("="*80)
print("STEP 5: AGGREGATE US SALES TO YEARLY (FIXED)")
print("="*80)

# -------------------------------------------------------------------
# 0. Normalize DIMENSION tables FIRST (CRITICAL FIX)
# -------------------------------------------------------------------

us_products_df = (
    us_products_df
    .withColumnRenamed("Product ID", "product_id")
    .withColumnRenamed("Product", "product_name")
)

us_states_df = (
    us_states_df
    .withColumnRenamed("State ID", "state_id")
    .withColumnRenamed("State", "state_name")
)

# -------------------------------------------------------------------
# 1. Aggregate US sales to yearly level
# -------------------------------------------------------------------

us_sales_yearly_df = (
    us_sales_df
    .withColumn("year", year(col("Purchase Date")))
    .groupBy(
        col("year"),
        col("Product ID"),
        col("State ID")
    )
    .agg(
        avg(col("Sale Price ($)")).alias("avg_sale_price"),
        stddev(col("Sale Price ($)")).alias("price_volatility"),
        min(col("Sale Price ($)")).alias("min_sale_price"),
        max(col("Sale Price ($)")).alias("max_sale_price"),
        sum(col("Quantity Sold (lb)")).alias("total_quantity_sold_lb"),
        avg(col("Quantity Sold (lb)")).alias("avg_quantity_sold_lb"),
        sum(col("Quantity Purchased (lb)")).alias("total_quantity_purchased_lb"),
        avg(col("Purchase Price ($)")).alias("avg_purchase_price"),
        count("*").alias("transaction_count")
    )
)

print(f"\n✓ US Sales Aggregated: {us_sales_yearly_df.count()} records")
us_sales_yearly_df.show(10)

# -------------------------------------------------------------------
# 2. Normalize FACT table column names
# -------------------------------------------------------------------

us_sales_yearly_df = (
    us_sales_yearly_df
    .withColumnRenamed("Product ID", "product_id")
    .withColumnRenamed("State ID", "state_id")
)

# -------------------------------------------------------------------
# 3. Enrich with product and state names (FIXED JOINS)
# -------------------------------------------------------------------

us_sales_enriched_df = (
    us_sales_yearly_df
    .join(us_products_df, on="product_id", how="left")
    .join(us_states_df, on="state_id", how="left")
    .select(
        col("year"),
        col("state_id"),
        col("state_name"),
        col("product_id"),
        col("product_name"),
        col("avg_sale_price"),
        col("price_volatility"),
        col("min_sale_price"),
        col("max_sale_price"),
        col("total_quantity_sold_lb"),
        col("avg_quantity_sold_lb"),
        col("total_quantity_purchased_lb"),
        col("avg_purchase_price"),
        col("transaction_count")
    )
)

print(f"\n✓ US Sales Enriched: {us_sales_enriched_df.count()} records")
us_sales_enriched_df.show(10, truncate=False)

# -------------------------------------------------------------------
# 4. Save to SILVER layer (correct medallion usage)
# -------------------------------------------------------------------

us_sales_enriched_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(
        "argisight_spark_workspace.silver.us_sales_yearly"
    )

print("\n✓ Saved to Delta table:")
print("argisight_spark_workspace.silver.us_sales_yearly")

# -------------------------------------------------------------------
# 5. Yearly summary sanity check
# -------------------------------------------------------------------

print("\n=== US Sales Yearly Summary ===")
(
    us_sales_enriched_df
    .groupBy("year")
    .agg(
        count("*").alias("records"),
        countDistinct("product_id").alias("unique_products"),
        countDistinct("state_id").alias("unique_states"),
        avg("avg_sale_price").alias("overall_avg_price")
    )
    .orderBy("year")
    .show()
)

print("="*80)
```

### Join US + FAOSTAT Data

```python
from pyspark.sql.functions import *

print("="*80)
print("STEP 6: INTEGRATE US + FAOSTAT DATA (FIXED)")
print("="*80)

# -------------------------------------------------------------------
# Load tables
# -------------------------------------------------------------------

us_yearly_df = spark.table(
    "argisight_spark_workspace.silver.us_sales_yearly"
).alias("us")

fo_yearly_df = spark.table(
    "argisight_spark_workspace.bronze.fo_usa_pivot_df"
).alias("fo")

product_map_df = spark.table(
    "argisight_spark_workspace.bronze.product_mapping_df"
).alias("pm")

print(f"\nInput Tables:")
print(f"  - US Sales Yearly: {us_yearly_df.count()} records")
print(f"  - FAOSTAT Yearly: {fo_yearly_df.count()} records")
print(f"  - Product Mapping: {product_map_df.count()} records")

# -------------------------------------------------------------------
# Step 1: Join US sales with product mapping
# -------------------------------------------------------------------

us_with_mapping_df = (
    us_yearly_df
    .join(
        product_map_df,
        col("us.product_id") == col("pm.us_product_id"),
        "left"
    )
)

print(f"\n✓ US with Product Mapping: {us_with_mapping_df.count()} records")

# -------------------------------------------------------------------
# Step 2: Join with FAOSTAT data
# -------------------------------------------------------------------

integrated_df = (
    us_with_mapping_df
    .join(
        fo_yearly_df,
        (col("pm.faostat_item_code") == col("fo.item_code")) &
        (col("us.year") == col("fo.year")),
        "left"
    )
)

print(f"\n✓ After FAOSTAT Join: {integrated_df.count()} records")

# -------------------------------------------------------------------
# Step 3: Select and organize final columns (NO AMBIGUITY)
# -------------------------------------------------------------------

integrated_final_df = integrated_df.select(
    col("us.year").alias("year"),
    col("us.state_id"),
    col("us.state_name"),
    col("us.product_id"),
    col("us.product_name").alias("us_product_name"),
    col("us.avg_sale_price"),
    col("us.price_volatility"),
    col("us.min_sale_price"),
    col("us.max_sale_price"),
    col("us.total_quantity_sold_lb"),
    col("us.avg_quantity_sold_lb"),
    col("us.total_quantity_purchased_lb"),
    col("us.avg_purchase_price"),
    col("us.transaction_count"),
    col("pm.faostat_item_code"),
    col("fo.item").alias("faostat_item_name"),
    col("fo.production_tonnes").alias("usa_production_tonnes"),
    col("fo.area_harvested_ha").alias("usa_area_harvested_ha"),
    col("fo.yield_kg_per_ha").alias("usa_yield_kg_per_ha"),
    col("pm.match_type")
)

print(f"\n✓ Integrated Final Dataset: {integrated_final_df.count()} records")
integrated_final_df.show(20, truncate=False)

# -------------------------------------------------------------------
# Save to SILVER layer
# -------------------------------------------------------------------

integrated_final_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(
        "argisight_spark_workspace.silver.integrated_us_faostat"
    )

print("\n✓ Saved to Delta table:")
print("argisight_spark_workspace.silver.integrated_us_faostat")
print("="*80)
```

##  PHASE 3: FEATURE ENGINEERING

### Time-Based Features

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
from pyspark.sql.types import *

print("="*80)
print("PHASE 3: FEATURE ENGINEERING - STEP 1")
print("Time-Based Features")
print("="*80)

# Load integrated data
integrated_df = spark.table("argisight_spark_workspace.gold.integrated_us_faostat")

# Create time-based features
features_time = integrated_df \
    .withColumn("month", lit(1)) \
    .withColumn("quarter", 
        when(col("year") % 4 == 0, 1)
        .when(col("year") % 4 == 1, 2)
        .when(col("year") % 4 == 2, 3)
        .otherwise(4)
    ) \
    .withColumn("year_index", col("year") - 2019) \
    .withColumn("is_recent_year", 
        when(col("year").isin([2022, 2023]), 1).otherwise(0)
    ) \
    .withColumn("is_harvest_season",
        when(col("us_product_name").isin([
            "Apples", "Grapes", "Corn", "Tomatoes", "Potatoes", "Carrots"
        ]), 1).otherwise(0)
    ) \
    .withColumn("years_since_start", 
        col("year") - lit(2019)
    )

print(f"\n✓ Created 6 time-based features")
print("  • month (always 1 for yearly data)")
print("  • quarter (1-4)")
print("  • year_index (0-4)")
print("  • is_recent_year (binary)")
print("  • is_harvest_season (binary)")
print("  • years_since_start (0-4)")

features_time.select("year", "us_product_name", "quarter", "year_index", 
                     "is_harvest_season", "years_since_start").show(10)

print("="*80)
```

### FAOSTAT-Derived Features

```python
print("="*80)
print("PHASE 3: FEATURE ENGINEERING - STEP 2")
print("FAOSTAT-Derived Features")
print("="*80)

# Window for year-over-year calculations
window_product = Window.partitionBy("product_id").orderBy("year")

# FAOSTAT production features
features_faostat = features_time \
    .withColumn("production_lag_1y",
        lag("usa_production_tonnes", 1).over(window_product)
    ) \
    .withColumn("production_yoy_growth",
        when(col("production_lag_1y").isNotNull() & (col("production_lag_1y") != 0),
             (col("usa_production_tonnes") - col("production_lag_1y")) / col("production_lag_1y") * 100
        ).otherwise(None)
    ) \
    .withColumn("yield_efficiency_ratio",
        when(col("usa_yield_kg_per_ha").isNotNull() & (col("usa_yield_kg_per_ha") > 0),
             col("usa_production_tonnes") / (col("usa_area_harvested_ha") * col("usa_yield_kg_per_ha") / 1000)
        ).otherwise(None)
    ) \
    .withColumn("production_per_area",
        when(col("usa_area_harvested_ha").isNotNull() & (col("usa_area_harvested_ha") > 0),
             col("usa_production_tonnes") / col("usa_area_harvested_ha")
        ).otherwise(None)
    ) \
    .withColumn("production_intensity_index",
        when(col("usa_production_tonnes").isNotNull(),
             col("usa_production_tonnes") / 1000000  # Normalize to millions
        ).otherwise(None)
    )

print(f"\n✓ Created 5 FAOSTAT-derived features")
print("  • production_lag_1y (tonnes)")
print("  • production_yoy_growth (%)")
print("  • yield_efficiency_ratio")
print("  • production_per_area (tonnes/ha)")
print("  • production_intensity_index")

# Show examples for products with FAOSTAT data
features_faostat.filter(col("usa_production_tonnes").isNotNull()) \
    .select("year", "us_product_name", "usa_production_tonnes", 
            "production_yoy_growth", "production_per_area") \
    .orderBy("us_product_name", "year") \
    .show(10, truncate=False)

print("="*80)
```

### Price-Based Features

```python
print("="*80)
print("PHASE 3: FEATURE ENGINEERING - STEP 3")
print("Price-Based Features")
print("="*80)

# Window for product-level statistics
window_product_state = Window.partitionBy("product_id", "state_id").orderBy("year")
window_product_all = Window.partitionBy("product_id")

features_price = features_faostat \
    .withColumn("price_lag_1y",
        lag("avg_sale_price", 1).over(window_product_state)
    ) \
    .withColumn("price_yoy_change",
        when(col("price_lag_1y").isNotNull() & (col("price_lag_1y") != 0),
             (col("avg_sale_price") - col("price_lag_1y")) / col("price_lag_1y") * 100
        ).otherwise(None)
    ) \
    .withColumn("price_volatility_index",
        when(col("price_volatility").isNotNull() & (col("avg_sale_price") > 0),
             col("price_volatility") / col("avg_sale_price") * 100
        ).otherwise(None)
    ) \
    .withColumn("profit_margin_pct",
        when((col("avg_purchase_price").isNotNull()) & (col("avg_purchase_price") > 0),
             (col("avg_sale_price") - col("avg_purchase_price")) / col("avg_purchase_price") * 100
        ).otherwise(None)
    ) \
    .withColumn("price_zscore",
        (col("avg_sale_price") - avg("avg_sale_price").over(window_product_all)) / 
        stddev("avg_sale_price").over(window_product_all)
    )

print(f"\n✓ Created 5 price-based features")
print("  • price_lag_1y ($)")
print("  • price_yoy_change (%)")
print("  • price_volatility_index (%)")
print("  • profit_margin_pct (%)")
print("  • price_zscore")

features_price.select("year", "us_product_name", "state_name",
                      "avg_sale_price", "price_yoy_change", 
                      "profit_margin_pct", "price_zscore") \
    .orderBy("us_product_name", "year") \
    .show(10, truncate=False)

print("="*80)
```

### Rolling Window Features

```python
print("="*80)
print("PHASE 3: FEATURE ENGINEERING - STEP 4")
print("Rolling Window Features")
print("="*80)

# Rolling windows (2-year and 3-year)
window_2y = Window.partitionBy("product_id", "state_id").orderBy("year").rowsBetween(-1, 0)
window_3y = Window.partitionBy("product_id", "state_id").orderBy("year").rowsBetween(-2, 0)
window_prod_2y = Window.partitionBy("product_id").orderBy("year").rowsBetween(-1, 0)

features_rolling = features_price \
    .withColumn("rolling_avg_price_2y",
        avg("avg_sale_price").over(window_2y)
    ) \
    .withColumn("rolling_std_price_2y",
        stddev("avg_sale_price").over(window_2y)
    ) \
    .withColumn("rolling_avg_volume_3y",
        avg("total_quantity_sold_lb").over(window_3y)
    ) \
    .withColumn("rolling_avg_production_2y",
        when(col("usa_production_tonnes").isNotNull(),
             avg("usa_production_tonnes").over(window_prod_2y)
        ).otherwise(None)
    )

print(f"\n✓ Created 4 rolling window features")
print("  • rolling_avg_price_2y (2-year moving average)")
print("  • rolling_std_price_2y (2-year moving std)")
print("  • rolling_avg_volume_3y (3-year volume avg)")
print("  • rolling_avg_production_2y (2-year production avg)")

features_rolling.select("year", "us_product_name", "state_name",
                        "avg_sale_price", "rolling_avg_price_2y", 
                        "rolling_std_price_2y") \
    .orderBy("us_product_name", "state_name", "year") \
    .show(10, truncate=False)

print("="*80)
```

### Interaction Features

```python
print("="*80)
print("PHASE 3: FEATURE ENGINEERING - STEP 5")
print("Interaction Features")
print("="*80)

# Interaction features combining multiple dimensions
features_interaction = features_rolling \
    .withColumn("price_per_production_ratio",
        when((col("usa_production_tonnes").isNotNull()) & (col("usa_production_tonnes") > 0),
             col("avg_sale_price") * col("total_quantity_sold_lb") / (col("usa_production_tonnes") * 2204.62)
        ).otherwise(None)
    ) \
    .withColumn("demand_supply_index",
        when((col("usa_production_tonnes").isNotNull()) & (col("usa_production_tonnes") > 0),
             col("total_quantity_sold_lb") / (col("usa_production_tonnes") * 2204.62) * 100
        ).otherwise(None)
    ) \
    .withColumn("price_volume_interaction",
        col("avg_sale_price") * log1p(col("total_quantity_sold_lb"))
    )

print(f"\n✓ Created 3 interaction features")
print("  • price_per_production_ratio")
print("  • demand_supply_index (state sales / national production %)")
print("  • price_volume_interaction")

features_interaction.filter(col("usa_production_tonnes").isNotNull()) \
    .select("year", "us_product_name", "state_name",
            "price_per_production_ratio", "demand_supply_index") \
    .orderBy("us_product_name", "year") \
    .show(10, truncate=False)

print("="*80)
```

### Categorical Encoding

```python
print("="*80)
print("PHASE 3: FEATURE ENGINEERING - STEP 6")
print("Categorical Encoding")
print("="*80)

# Create categorical indicators
features_categorical = features_interaction \
    .withColumn("is_fruit",
        when(col("us_product_name").isin([
            "Apples", "Oranges", "Blueberries", "Strawberries", "Peaches",
            "Grapes", "Watermelons", "Avocados"
        ]), 1).otherwise(0)
    ) \
    .withColumn("is_vegetable",
        when(col("us_product_name").isin([
            "Corn", "Tomatoes", "Lettuce", "Potatoes", "Carrots",
            "Spinach", "Bell Peppers"
        ]), 1).otherwise(0)
    ) \
    .withColumn("is_herb",
        when(col("match_type") == "No Match", 1).otherwise(0)
    )

print(f"\n✓ Created 3 categorical features")
print("  • is_fruit (binary)")
print("  • is_vegetable (binary)")
print("  • is_herb (binary)")

features_categorical.groupBy("us_product_name") \
    .agg(
        first("is_fruit").alias("is_fruit"),
        first("is_vegetable").alias("is_vegetable"),
        first("is_herb").alias("is_herb")
    ).show(22, truncate=False)

print("="*80)
```

### Outlier Detection

```python
print("="*80)
print("STEP 7: Outlier Detection Analysis")
print("="*80)

from scipy import stats

# Calculate outliers using IQR method
def find_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] < lower_bound) | (data[column] > upper_bound)]

outlier_features = ['avg_sale_price', 'price_yoy_change', 'production_yoy_growth']

print("\n=== OUTLIER SUMMARY ===")
outlier_counts = {}
for feature in outlier_features:
    outliers = find_outliers(features_pd, feature)
    outlier_counts[feature] = len(outliers)
    print(f"\n{feature}:")
    print(f"  • Outlier count: {len(outliers)} ({len(outliers)/len(features_pd)*100:.1f}%)")
    print(f"  • Values: {outliers[feature].min():.2f} to {outliers[feature].max():.2f}")

# Visualize outliers
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Outlier Detection (Box Plots)', fontsize=14, fontweight='bold')

for idx, feature in enumerate(outlier_features):
    axes[idx].boxplot([features_pd[feature].dropna()], labels=[feature])
    axes[idx].set_title(feature, fontweight='bold')
    axes[idx].set_ylabel('Value')
    axes[idx].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('/dbfs/FileStore/agrisight/eda/06_outliers.png', dpi=300, bbox_inches='tight')
print("✓ Saved: /dbfs/FileStore/agrisight/eda/06_outliers.png")
plt.show()
plt.close()

print("="*80)
```

### Create EDA Metadata Table

```python
from pyspark.sql import functions as F

print("="*80)
print("STEP 12: Save EDA Metadata")
print("="*80)

feature_stats_data = []

for feature in numeric_features:
    data = features_pd[feature].dropna()
    if len(data) > 0:
        feature_stats_data.append({
            'feature_name': feature,
            'count': int(len(data)),
            'mean': float(data.mean()),
            'std': float(data.std()),
            'min': float(data.min()),
            'max': float(data.max()),
            'median': float(data.median()),
            'q25': float(data.quantile(0.25)),
            'q75': float(data.quantile(0.75))
        })

eda_stats_df = spark.createDataFrame(feature_stats_data)
eda_stats_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("argisight_spark_workspace.metadata.eda_feature_statistics")


print("✓ Feature statistics saved to: metadata.eda_feature_statistics")

correlations_data = []

for feature in numeric_features:
    try:
        corr_val = importance_df[feature].corr(features_pd['avg_sale_price'])
        if corr_val is not None:
            correlations_data.append({
                'feature_name': feature,
                'correlation_with_price': float(corr_val),
                'abs_correlation': float(abs(corr_val))
            })
    except Exception:
        pass

# Rank by importance
correlations_data = sorted(
    correlations_data,
    key=lambda x: x['abs_correlation'],
    reverse=True
)

for idx, row in enumerate(correlations_data, start=1):
    row['importance_rank'] = idx

# corr_insights_df = spark.createDataFrame(correlations_data)
if len(correlations_data) > 0:
    corr_insights_df = spark.createDataFrame(correlations_data)

    corr_insights_df.write \
        .format("delta") \
        .mode("overwrite") \
        .option("overwriteSchema", "true") \
        .saveAsTable(
            "argisight_spark_workspace.metadata.eda_feature_correlations"
        )

    print("✓ Feature correlations saved to: metadata.eda_feature_correlations")
else:
    print("⚠ No valid correlations found — table not created")


```
